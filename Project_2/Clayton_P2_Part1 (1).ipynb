## 6 ##

```python
# Number 6
def classify_unemployment(unemployment_rate):
    if unemployment_rate < 4.10:
        return 'low'
    elif unemployment_rate < 5.3:
        return 'moderate'
    else:
        return 'high'

df_loaded['unemployment_rank'] = df_loaded['unemployment_rate'].apply(classify_unemployment)
df_loaded

# Created a new column called "unemployment rank" which filtered unemployment rates into groups that categorizes it.
```

The new column added entitled "unemployment_rank" specifically categorizes the unemployment rate of the row into three seperate categories: low, moderate, or high. Meaning, we are able to assess if the state had a low, moderate, or high rate of unemployment per month of measurement. This categorization can further be used to analyze relationships with other variables and group the data such as what is the averge rate crime when the dataset is grouped by unemployment_rank. So for all months when the unemployment rate was categorized as high, what was the crime rate. Allowing us to answer questions on what the unemplyment can influence.
